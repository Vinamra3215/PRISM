```python
import os
import sys
import pandas as pd
import numpy as np
import plotly.graph_objects as go

sys.path.append(os.path.abspath(os.path.join('..')))

feature_data_path = os.path.join("..", "data", "processed", "smoothed_features.parquet")
df = pd.read_parquet(feature_data_path)
total_days = len(df)

start_day = int(input(f"Enter START day number (1 to {total_days}): "))
end_day = int(input(f"Enter END day number ({start_day} to {total_days}): "))

days_vector = np.arange(start_day, end_day + 1)
df_filtered = df[df['Day_Number'].isin(days_vector)]

# Generate Momentum Oscillation Canvas
fig = go.Figure()

# Structural RSI Index Track
fig.add_trace(go.Scatter(x=df_filtered['Day_Number'], y=df_filtered['RSI'], name="Relative Strength Index", line=dict(color="mediumseagreen", width=2.5)))

# Add horizontal boundary threshold lines
fig.add_hline(y=70, line_dash="dash", line_color="crimson", annotation_text="Overbought Ceiling (70)")
fig.add_hline(y=30, line_dash="dash", line_color="royalblue", annotation_text="Oversold Floor (30)")

fig.update_layout(
    title=f"RSI Momentum Regime Sandbox (Days {start_day} to {end_day})",
    xaxis_title="Day Number",
    yaxis_title="Oscillator Index Scale (0 - 100)",
    yaxis=dict(range=[0, 100]),
    hovermode="x unified",
    template="plotly_white"
)
fig.show()
```
