```python
import os
import sys
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Route to parent folder to ensure local project discovery
sys.path.append(os.path.abspath(os.path.join('..')))

# Load pre-processed metrics from cache
feature_data_path = "/home/soq/__shutupandbendover/devansh-swc-wala/smoothed_features.parquet"
df = pd.read_parquet(feature_data_path)
total_days = len(df)

# Interactive Day Vector Selector Prompt
start_day = int(input(f"Enter START day number (1 to {total_days}): "))
end_day = int(input(f"Enter END day number ({start_day} to {total_days}): "))

days_vector = np.arange(start_day, end_day + 1)
df_filtered = df[df['Day_Number'].isin(days_vector)]

# Generate Dual Y-Axis System Risk Chart
fig = make_subplots(specs=[[{"secondary_y": True}]])

fig.add_trace(
    go.Scatter(x=df_filtered['Day_Number'], y=df_filtered['Beta'], name="Rolling Beta vs Nifty 50", line=dict(color="darkorange", width=2.5)),
    secondary_y=False
)

fig.add_trace(
    go.Scatter(x=df_filtered['Day_Number'], y=df_filtered['Alpha'], name="Alpha Edge", line=dict(color="purple", width=2, dash="dot")),
    secondary_y=True
)


fig.update_layout(
    title=f"Beta & Alpha Risk Isolation Sandbox (Days {start_day} to {end_day})",
    xaxis_title="Day Number",
    hovermode="x unified",
    template="plotly_white"
)

fig.update_yaxes(title_text="<b>Beta Coefficient (Sensitivity)</b>", secondary_y=False)
fig.update_yaxes(title_text="<b>Alpha Value (Outperformance Return)</b>", secondary_y=True)
fig.show()
```
